```python

import pandas as pd
```

```python
df1 = pd.DataFrame(columns=['a','b','c'])
df2 = pd.DataFrame(columns=['a','b','c','d','e'])

df1['a'] = [1,2,3]
df1['b'] = [4,5,6]
df1['c'] = [7,8,9]


merged_df = pd.concat([df2,df1],axis=1)
# remove duplicate columns by keeping the last occurrence
merged_df = merged_df.loc[:, ~merged_df.columns.duplicated(keep='last')]
merged_df
```
