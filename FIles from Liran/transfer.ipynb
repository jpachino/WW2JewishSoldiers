```python

# Read the data from master.xlsx
master_path = datapath + r'\\NewMaster_NY.xlsx'

master_cols = ['Last Name','First Name','Country','Rank','Hometown','Type of Death','Fighting Description','Updated Medals']

master_cols
```

```python
# original
original_path = datapath + r'\\original.xlsx'

# Read the template from upload.xlsx
upload_path = datapath + r'\\upload_NY.xlsx'
upload_cols = ['LNameEn','FNameEn','BirthCountry','DegreeEn', 'BirthCityEn','DeathDetailsEn','FightingDescEn', 'MedalEn']


mapping = {key:value for key,value in zip(master_cols,upload_cols)} 
# Read all sheets from master.xlsx with the correct header row
master_sheets = pd.read_excel(master_path, sheet_name=None, header=0)

mapping
```

```python


medals_dict = {
  "BSM": "Bronze Star Medal",
  "PH": "Purple Heart",
  "SS": "Silver Star",
  "DFC": "Distinguished Flying Cross",
  "OLC": "Oak Leaf Cluster",
  "AM": "Air Medal",
  "PUC": "Presidential Unit Citation",
  "COMM": "Certificate of Merit Medal",
}

fighting_dict = {
  "A" : "Army",
  "N" : "Navy",
  "M" : "Marine Corps",
}

```

```python

# prompt: for all in master_sheets: 
# -create a Country column and put "6180" in all rows.
# -create a 'Fighting Description' column and if  'Awards/Medals' first char include in fighting_dict, move its value in the dict to  'Fighting Description' column


# ... (Your existing code for imports, paths, and dictionaries)

for sheet_name, df in master_sheets.items():
    # Create 'Country' column and fill with "6180"
    df['Country'] = '6180'
    
    # Create 'Fighting Description' column and populate based on 'Awards/Medals'
    df['Fighting Description'] = ''  # Initialize with empty strings

    for index, row in df.iterrows():
        awards = str(row['Awards/Medals'])
        if awards and awards[0] in fighting_dict:
            df.loc[index, 'Fighting Description'] = fighting_dict[awards[0]]
            df.loc[index, 'Awards/Medals'] = ''  # Clear the 'Awards/Medals' column

    # Save the modified DataFrame back to the Excel file (optional)
    # This will overwrite the original sheet
    # with pd.ExcelWriter(master_path, engine='openpyxl', mode='a') as writer:
    #   df.to_excel(writer, sheet_name=sheet_name, index=False)
    
    master_sheets[sheet_name] = df

# Example access of a modified sheet
# print(master_sheets['Sheet1']) # Replace 'Sheet1' with your actual sheet name
```

```python
for sheet_name, df in master_sheets.items():
    df['Updated Medals'] = ''
    for index, row in df.iterrows():
        updated_medals = []
        for col in ['Awards/Medals','Unnamed: 10','Unnamed: 11','Unnamed: 12','Unnamed: 13','Unnamed: 14']:
            value = str(row[col]).strip()
            if value in medals_dict:
                updated_medals.append(medals_dict[value])
            elif value:
                updated_medals.append(value)
        df.loc[index, 'Updated Medals'] = ', '.join(updated_medals)
    master_sheets[sheet_name] = df
```
